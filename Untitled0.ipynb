```python
from torchvision import datasets

data = datasets.OxfordIIITPet(
    root="./data",
    split="trainval",
    target_types="category",
    download=True
)

print("数据集下载完成！")
print("图片数量：", len(data))
print("类别数量：", len(data.classes))
```

```python
from torchvision import datasets
from sklearn.model_selection import train_test_split
import numpy as np

# 加载训练+验证数据
trainval_data = datasets.OxfordIIITPet(
    root="./data",
    split="trainval",
    target_types="category",
    download=False
)

# 加载官方测试数据
test_data = datasets.OxfordIIITPet(
    root="./data",
    split="test",
    target_types="category",
    download=False
)

# 合并所有数据的标签
all_labels = np.concatenate([
    np.array(trainval_data._labels),
    np.array(test_data._labels)
])

all_indices = np.arange(len(all_labels))

# 第一次：70%训练，30%临时数据
train_idx, temp_idx = train_test_split(
    all_indices,
    test_size=0.30,
    stratify=all_labels,
    random_state=42
)

# 第二次：把剩下30%平均分成15%验证、15%测试
val_idx, test_idx = train_test_split(
    temp_idx,
    test_size=0.50,
    stratify=all_labels[temp_idx],
    random_state=42
)

print("总数据量：", len(all_indices))
print("训练集：", len(train_idx))
print("验证集：", len(val_idx))
print("测试集：", len(test_idx))
```

```python
from torchvision import transforms

# 训练集：加入随机增强
train_transform = transforms.Compose([
    transforms.Resize(256),
    transforms.RandomCrop(224),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

# 验证集和测试集：不进行随机增强
eval_transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

print("图片预处理设置完成！")
```

```python
import torch
import torchvision.models as models
from torch import nn

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = models.resnet18(weights="DEFAULT")

# 把最后的分类层改成37类
model.fc = nn.Linear(model.fc.in_features, 37)

model = model.to(device)

print("模型创建完成！")
print("使用设备：", device)
print("分类类别数：", model.fc.out_features)
```

```python
from torch.utils.data import Dataset, DataLoader
from PIL import Image
import os

class PetDataset(Dataset):
    def __init__(self, base_dataset, indices, transform):
        self.base_dataset = base_dataset
        self.indices = indices
        self.transform = transform

    def __len__(self):
        return len(self.indices)

    def __getitem__(self, idx):
        real_idx = self.indices[idx]
        image, label = self.base_dataset[real_idx]

        if self.transform:
            image = self.transform(image)

        return image, label


# 合并后的完整数据集
all_data = torch.utils.data.ConcatDataset([
    trainval_data,
    test_data
])

train_dataset = PetDataset(all_data, train_idx, train_transform)
val_dataset = PetDataset(all_data, val_idx, eval_transform)
test_dataset = PetDataset(all_data, test_idx, eval_transform)

train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True,
    num_workers=2
)

val_loader = DataLoader(
    val_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=2
)

test_loader = DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=2
)

print("数据加载器创建完成！")
print("训练批次数：", len(train_loader))
print("验证批次数：", len(val_loader))
print("测试批次数：", len(test_loader))
```

```python
import torch.optim as optim

# 损失函数
criterion = nn.CrossEntropyLoss()

# AdamW优化器
optimizer = optim.AdamW(
    model.parameters(),
    lr=1e-4,
    weight_decay=1e-4
)

print("损失函数和优化器设置完成！")
print("学习率：", 1e-4)
print("权重衰减：", 1e-4)
```

```python
def train_one_epoch(model, loader, criterion, optimizer, device):
    model.train()

    total_loss = 0
    correct = 0
    total = 0

    for images, labels in loader:
        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        total_loss += loss.item() * images.size(0)

        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()

    avg_loss = total_loss / total
    accuracy = correct / total

    return avg_loss, accuracy


def evaluate(model, loader, criterion, device):
    model.eval()

    total_loss = 0
    correct = 0
    total = 0

    with torch.no_grad():
        for images, labels in loader:
            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = criterion(outputs, labels)

            total_loss += loss.item() * images.size(0)

            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct += predicted.eq(labels).sum().item()

    avg_loss = total_loss / total
    accuracy = correct / total

    return avg_loss, accuracy


print("训练和验证函数准备完成！")
```

```python
num_epochs = 10

train_losses = []
train_accs = []
val_losses = []
val_accs = []

best_val_acc = 0.0

for epoch in range(num_epochs):

    train_loss, train_acc = train_one_epoch(
        model, train_loader, criterion, optimizer, device
    )

    val_loss, val_acc = evaluate(
        model, val_loader, criterion, device
    )

    train_losses.append(train_loss)
    train_accs.append(train_acc)
    val_losses.append(val_loss)
    val_accs.append(val_acc)

    print(
        f"Epoch [{epoch+1}/{num_epochs}] "
        f"Train Loss: {train_loss:.4f} "
        f"Train Acc: {train_acc:.4f} "
        f"Val Loss: {val_loss:.4f} "
        f"Val Acc: {val_acc:.4f}"
    )

    # 保存验证集表现最好的模型
    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(model.state_dict(), "baseline_best.pth")

print("Baseline训练完成！")
print("最佳验证集准确率：", best_val_acc)
```

```python
from sklearn.metrics import accuracy_score, f1_score
import numpy as np

# 加载验证集表现最好的模型
model.load_state_dict(torch.load("baseline_best.pth", map_location=device))
model.eval()

all_preds = []
all_labels = []

with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)

        outputs = model(images)
        preds = outputs.argmax(dim=1)

        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.numpy())

# Top-1 Accuracy
top1_acc = accuracy_score(all_labels, all_preds)

# Macro-F1
macro_f1 = f1_score(
    all_labels,
    all_preds,
    average="macro"
)

print("Baseline测试集结果")
print("Top-1 Accuracy：", top1_acc)
print("Macro-F1：", macro_f1)
```

```python
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# 计算混淆矩阵
cm = confusion_matrix(all_labels, all_preds)

# 绘制混淆矩阵
plt.figure(figsize=(14, 12))
sns.heatmap(
    cm,
    cmap="Blues",
    xticklabels=data.classes,
    yticklabels=data.classes
)

plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.title("Confusion Matrix - Baseline")
plt.tight_layout()
plt.show()
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision.models as models

# 使用 GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 创建改进版 ResNet-18
model_improved = models.resnet18(weights="DEFAULT")

# 修改最后一层为 37 个类别
model_improved.fc = nn.Linear(
    model_improved.fc.in_features,
    37
)

model_improved = model_improved.to(device)

# 损失函数
criterion_improved = nn.CrossEntropyLoss()

# 优化器
optimizer_improved = optim.AdamW(
    model_improved.parameters(),
    lr=1e-4,
    weight_decay=1e-4
)

print("改进版模型准备完成！")
print("设备：", device)
print("类别数：", model_improved.fc.out_features)
print("损失函数：", criterion_improved)
print("优化器：", optimizer_improved)
```

```python
from torch.utils.data import DataLoader

train_dataset_improved = PetDataset(
    all_data,
    train_idx,
    train_transform_improved
)

train_loader_improved = DataLoader(
    train_dataset_improved,
    batch_size=32,
    shuffle=True,
    num_workers=2
)

print("改进版训练数据加载器创建完成！")
print("训练图片数量：", len(train_dataset_improved))
print("训练批次数：", len(train_loader_improved))
```

```python
num_epochs = 10

train_losses_improved = []
val_losses_improved = []
train_accs_improved = []
val_accs_improved = []

best_val_acc_improved = 0.0

for epoch in range(num_epochs):

    # ===== 训练 =====
    model_improved.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in train_loader_improved:

        images = images.to(device)
        labels = labels.to(device)

        optimizer_improved.zero_grad()

        outputs = model_improved(images)

        loss = criterion_improved(outputs, labels)

        loss.backward()

        optimizer_improved.step()

        running_loss += loss.item() * images.size(0)

        _, predicted = torch.max(outputs, 1)

        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    train_loss = running_loss / total
    train_acc = correct / total


    # ===== 验证 =====
    model_improved.eval()

    val_loss_total = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():

        for images, labels in val_loader:

            images = images.to(device)
            labels = labels.to(device)

            outputs = model_improved(images)

            loss = criterion_improved(outputs, labels)

            val_loss_total += loss.item() * images.size(0)

            _, predicted = torch.max(outputs, 1)

            val_total += labels.size(0)
            val_correct += (predicted == labels).sum().item()


    val_loss = val_loss_total / val_total
    val_acc = val_correct / val_total


    train_losses_improved.append(train_loss)
    val_losses_improved.append(val_loss)
    train_accs_improved.append(train_acc)
    val_accs_improved.append(val_acc)


    print(
        f"Epoch [{epoch+1}/{num_epochs}] "
        f"Train Acc: {train_acc:.4f} "
        f"Val Acc: {val_acc:.4f}"
    )


    # 保存最好模型
    if val_acc > best_val_acc_improved:
        best_val_acc_improved = val_acc

        torch.save(
            model_improved.state_dict(),
            "improved_best.pth"
        )


print("改进版训练完成！")
print(
    f"最佳验证准确率: {best_val_acc_improved:.4f}"
)
```

```python
from sklearn.metrics import accuracy_score, f1_score

# 加载最佳改进模型
model_improved.load_state_dict(
    torch.load("improved_best.pth", map_location=device)
)

model_improved.eval()

preds_improved = []
labels_improved = []

with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)

        outputs = model_improved(images)

        preds = outputs.argmax(dim=1)

        preds_improved.extend(preds.cpu().numpy())
        labels_improved.extend(labels.numpy())


top1_improved = accuracy_score(
    labels_improved,
    preds_improved
)

macro_f1_improved = f1_score(
    labels_improved,
    preds_improved,
    average="macro"
)

print("改进版测试结果")
print("Top-1 Accuracy:", top1_improved)
print("Macro-F1:", macro_f1_improved)
```
